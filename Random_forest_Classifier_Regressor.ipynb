**Regressor**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import category_encoders as ce

# 1. Map target categorical values to continuous/numerical values
# (Adjust order if your classes are defined differently)
class_mapping = {'unacc': 0, 'acc': 1, 'good': 2, 'vgood': 3}
y_reg = y.map(class_mapping)

# 2. Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y_reg, test_size=0.25, random_state=42
)

# 3. Fit Ordinal Encoder (Important: fit on X_train only, then transform X_test)
encoder = ce.OrdinalEncoder(
    cols=['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety']
)
X_train_encoded = encoder.fit_transform(X_train)
X_test_encoded = encoder.transform(X_test)

# 4. Instantiate and fit RandomForestRegressor
rfr = RandomForestRegressor(n_estimators=200, random_state=42)
rfr.fit(X_train_encoded, y_train)

# 5. Predict and evaluate regression metrics
y_pred = rfr.predict(X_test_encoded)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R² Score: {r2:.4f}")
```
