**Support Vector Regression (SVR)**

 supervised machine learning algorithm used to predict continuous values rather than categorical labels.Unlike traditional linear regression—which minimizes the sum of squared errors—SVR attempts to fit the best-fitting line or hyperplane within an $\epsilon$-insensitive tube ($\epsilon$-margin) around the data.

```python
#independent and dependent features
X= df[[ 'tip', 'sex', 'smoker', 'day', 'time', 'size']]
y= df['total_bill']
```

```python
##train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
X_train.head()
```

```python
#feature encoding
#label
#one hot
from sklearn.preprocessing import LabelEncoder
```

```python
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
import warnings
warnings.filterwarnings('ignore')
X_train['sex']=le1.fit_transform(X_train['sex'])
X_train['smoker']=le2.fit_transform(X_train['smoker'])
X_train['time']=le3.fit_transform(X_train['time'])
```

```python
X_train.head()
```

```python
import warnings
warnings.filterwarnings('ignore')
X_test['sex']=le1.transform(X_test['sex'])
X_test['smoker']=le2.transform(X_test['smoker'])
X_test['time']=le3.transform(X_test['time'])
```

```python
X_test.head()
```

```python
#one hot encoding -- column transformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)
X_train=ct.fit_transform(X_train)
```

```python
X_test=ct.transform(X_test)
```

```python
X_test
```

```python
#apply svm
from sklearn.svm import SVR
svr = SVR()
```

```python
svr.fit(X_train,y_train)
```

```python
y_pred = svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```

```python
##Hyperparameter using gridsearchcv
from sklearn.model_selection import GridSearchCV

#define parameter range
param_grid = {'C': [0.1,1,10,100,1000],
              'gamma':[1,0.1,0.01,0.001,0.0001],
              'kernel':['rbf']}
```

```python
grid = GridSearchCV(SVR(),param_grid,refit=True,verbose=2)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,grid_pred))
print(mean_absolute_error(y_test,grid_pred))
```
